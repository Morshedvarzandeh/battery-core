# Transport module — Fick's first law: which way does lithium move?

This notebook is an existing physics building block, separate from the numbered fundamentals sequence. It uses the tested `battery_core` package as the single source of truth.

For one-dimensional diffusion, \(J=-D\,dc/dx\). Here \(D\) is in m²/s, \(dc/dx\) is in mol/m⁴, and \(J\) is in mol/(m²·s). The minus sign means diffusion proceeds down the concentration gradient.

```python
import matplotlib.pyplot as plt
import numpy as np
from battery_core import ficks_first_law_flux
```

## Scalar example

```python
D = 1.0e-14
gradient = 2.0e6
flux = ficks_first_law_flux(D, gradient)
print(f'J = {flux:.2e} mol/(m² s)')
```

## Vectorized example and learner control

```python
learner_diffusivity = 1.0e-14
gradients = np.linspace(-3.0e6, 3.0e6, 201)
fluxes = ficks_first_law_flux(learner_diffusivity, gradients)
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(gradients, fluxes)
ax.axhline(0.0, linewidth=0.8)
ax.axvline(0.0, linewidth=0.8)
ax.set(xlabel='Concentration gradient [mol m⁻⁴]', ylabel='Molar flux [mol m⁻² s⁻¹]', title=f"Fick's first law, D={learner_diffusivity:.1e} m²/s")
ax.grid(alpha=0.3)
plt.show()
```

## Representative battery-material example

Using an illustrative solid diffusivity of \(10^{-14}\) m²/s, a concentration change of -5000 mol/m³ over 5 µm gives a gradient of \(-10^9\) mol/m⁴.

```python
gradient_lfp = -5000.0 / 5.0e-6
flux_lfp = ficks_first_law_flux(1.0e-14, gradient_lfp)
print(f'Gradient = {gradient_lfp:.2e} mol/m⁴')
print(f'Flux = {flux_lfp:.2e} mol/(m² s)')
```

## Assumptions and limitations

This is a local one-dimensional constitutive relation, not a transient cell model. It does not define a mesh, initial conditions, boundary conditions, reactions, migration, mechanics, or porous-electrode homogenization.
